```python
# Step 3: Launch Headless FastAPI Server in Background
import subprocess
import time

# Start uvicorn server
server_proc = subprocess.Popen([
    'python3', '-m', 'uvicorn',
    'tools.cloud_worker.colab_server:app',
    '--host', '0.0.0.0',
    '--port', '8000'
])
time.sleep(2)
print('✅ FastAPI server running on port 8000!')
```
